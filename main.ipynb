```python
from dolfinx import mesh, fem, la
from mpi4py import MPI
from ufl import TrialFunction, TestFunction, grad, inner, dx
import ufl

# Create custom rectangular mesh (e.g., 3 x 2 domain)
domain = mesh.create_rectangle(
    comm=MPI.COMM_WORLD,
    points=[(0.0, 0.0), (3.0, 2.0)],
    n=[60, 40],
    cell_type=mesh.CellType.triangle
)
# Define the finite element using UFL
# element = ufl.FiniteElement("Lagrange", domain.ufl_cell(), 1)

cell_type = domain.topology.cell_type.name  # Get cell name as string

# Create vector element
element = basix.ufl.element("Lagrange", cell_type, 1, shape=(2,))




# Define function space on this new domain
V = fem.FunctionSpace(domain, element)

# Define trial/test functions
u = TrialFunction(V)
v = TestFunction(V)

# Define variational form
a = fem.form(inner(grad(u), grad(v))*dx)
L = fem.form(fem.Constant(domain, 0.0)*v*dx)

# Assembly and solve (as before)
A = fem.petsc.assemble_matrix(a)
A.assemble()
b = fem.petsc.assemble_vector(L)

solver = la.petsc.KSP().create(domain.comm)
solver.setOperators(A)

# Solution function
uh = fem.Function(V)
solver.solve(b, uh.vector)
```
